## NumPy Quick Implementation

```python
import numpy as np

rng = np.random.default_rng(42)

a = np.arange(1, 13).reshape(3, 4)
print(a)
print("shape:", a.shape)
print("row mean:", a.mean(axis=1))
print("column sum:", a.sum(axis=0))
print("values > 6:", a[a > 6])
```

## Pandas Quick Implementation

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E"],
    "Department": ["AI", "HR", "AI", "Finance", "AI"],
    "Experience": [1, 3, 5, 7, 9],
    "Salary": [35000, 42000, 60000, 65000, 80000]
})

print(df.head())
print("\nSummary:\n", df.describe())
print("\nAI employees:\n", df[df["Department"] == "AI"])
print("\nDepartment average salary:\n", df.groupby("Department")["Salary"].mean())
```

## Matplotlib Quick Implementation

```python
import matplotlib.pyplot as plt

plt.plot(df["Experience"], df["Salary"])
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Experience vs Salary")
plt.show()

plt.bar(df["Name"], df["Salary"])
plt.title("Salary by Employee")
plt.show()
```

## Seaborn Quick Implementation

```python
import seaborn as sns

sns.scatterplot(data=df, x="Experience", y="Salary", hue="Department")
plt.title("Experience vs Salary by Department")
plt.show()

sns.boxplot(data=df, x="Department", y="Salary", hue="Department", legend=False)
plt.title("Salary by Department")
plt.show()
```

## Mini Integrated Analysis

```python
numeric = df[["Experience", "Salary"]]
corr = numeric.corr()

print(corr)

sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
