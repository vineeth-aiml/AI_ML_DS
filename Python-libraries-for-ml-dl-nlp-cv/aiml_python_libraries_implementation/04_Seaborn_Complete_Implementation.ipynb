## 1. Import and Dataset

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "Name": list("ABCDEFGHIJKL"),
    "Department": ["AI", "HR", "AI", "Finance", "HR", "AI", "Finance", "AI", "Sales", "Sales", "AI", "HR"],
    "Experience": [1, 2, 3, 4, 5, 6, 7, 8, 2, 5, 9, 4],
    "Salary": [30000, 35000, 40000, 45000, 50000, 60000, 65000, 75000, 38000, 55000, 82000, 47000],
    "Age": [22, 23, 25, 27, 29, 30, 32, 35, 24, 31, 37, 28],
    "Performance": [3.1, 3.5, 3.8, 3.6, 4.0, 4.2, 4.1, 4.6, 3.4, 4.0, 4.8, 3.9]
})

df

default_color = plt.rcParams["axes.prop_cycle"].by_key()["color"][0]
```

## 2. Scatter Plot

```python
sns.scatterplot(data=df, x="Experience", y="Salary")
plt.title("Experience vs Salary")
plt.show()
```

## 3. Scatter Plot with hue

```python
sns.scatterplot(data=df, x="Experience", y="Salary", hue="Department")
plt.title("Experience vs Salary by Department")
plt.show()
```

## 5. Count Plot

```python
sns.countplot(data=df, x="Department", hue="Department")
plt.title("Employees by Department")
plt.show()
```

## 6. Bar Plot

```python
sns.barplot(data=df, x="Department", y="Salary", hue="Department")
plt.title("Average Salary by Department")
plt.show()
```

## 7. Histogram

```python
sns.histplot(data=df, x="Salary", bins=6, color=default_color)
plt.title("Salary Distribution")
plt.show()
```

## 8. Histogram with KDE

```python
sns.histplot(data=df, x="Salary", kde=True, color=default_color)
plt.title("Salary Distribution with KDE")
plt.show()
```

## 9. KDE Plot

```python
sns.kdeplot(data=df, x="Salary", fill=False, color=default_color)
plt.title("Salary Density")
plt.show()
```

## 10. Box Plot

```python
sns.boxplot(data=df, x="Department", y="Salary", hue="Department", legend=False)
plt.title("Salary by Department")
plt.show()
```

## 11. Violin Plot

```python
sns.violinplot(data=df, x="Department", y="Salary", hue="Department")
plt.title("Salary Distribution by Department")
plt.show()
```

## 12. Strip Plot

```python
sns.stripplot(data=df, x="Department", y="Salary", hue="Department")
plt.title("Individual Salary Observations")
plt.show()
```

## 13. Regression Plot

```python
sns.regplot(data=df, x="Experience", y="Salary", color=default_color)
plt.title("Experience vs Salary with Regression Line")
plt.show()
```

## 14. Correlation Matrix

```python
numeric_df = df[["Age", "Experience", "Salary", "Performance"]]
correlation = numeric_df.corr()

correlation
```

## 15. Heatmap

```python
sns.heatmap(correlation, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

## 16. Pair Plot

```python
sns.pairplot(
    df[["Age", "Experience", "Salary", "Performance"]],
    plot_kws={"color": default_color},
    diag_kws={"color": default_color}
)
plt.show()
```

## 17. Pair Plot with hue

```python
sns.pairplot(
    df[["Department", "Age", "Experience", "Salary", "Performance"]],
    hue="Department"
)
plt.show()
```

## 18. Joint Plot

```python
sns.jointplot(data=df, x="Experience", y="Salary", kind="scatter", color=default_color)
plt.show()
```

## 19. Categorical Plot (catplot)

```python
sns.catplot(data=df, x="Department", y="Salary", kind="box", hue="Department", legend=False)
plt.show()
```

## 20. Combining Seaborn with Matplotlib Controls

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df, x="Experience", y="Salary", hue="Department")
plt.xlabel("Years of Experience")
plt.ylabel("Monthly Salary")
plt.title("Employee Experience vs Salary")
plt.legend(title="Department")
plt.show()
```
