# NumPy Complete Implementation for AI/ML



## 1. Import NumPy

`np` is the standard alias. You may technically choose another valid alias, but `np` is the community convention.

```python
import numpy as np

print("NumPy version:", np.__version__)
```

## 2. Python List vs NumPy Array

NumPy arrays are designed for numerical computing and support vectorized operations.

```python
python_list = [1, 2, 3, 4]
numpy_array = np.array([1, 2, 3, 4])

print("Python list:", python_list)
print("NumPy array:", numpy_array)
print("Array type:", type(numpy_array))
```

## 3. Creating 1D, 2D and 3D Arrays

```python
a1 = np.array([10, 20, 30, 40])
a2 = np.array([[1, 2, 3], [4, 5, 6]])
a3 = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D:\n", a1)
print("\n2D:\n", a2)
print("\n3D:\n", a3)
```

## 4. Important Array Attributes

```python
for name, arr in [("a1", a1), ("a2", a2), ("a3", a3)]:
    print(
        name,
        "ndim =", arr.ndim,
        "shape =", arr.shape,
        "size =", arr.size,
        "dtype =", arr.dtype,
        "itemsize =", arr.itemsize,
        "bytes =", arr.nbytes
    )
```

## 5. Data Types and Type Conversion

```python
integers = np.array([1, 2, 3], dtype=np.int32)
floats = integers.astype(float)

print(integers, integers.dtype)
print(floats, floats.dtype)
```

## 6. Array Creation Helpers

```python
print("zeros:\n", np.zeros((2, 3)))
print("\nones:\n", np.ones((2, 3)))
print("\nfull:\n", np.full((2, 3), 7))
print("\nidentity:\n", np.eye(3))
print("\nempty (uninitialized values):\n", np.empty((2, 2)))
```

## 7. arange and linspace

```python
print("arange:", np.arange(1, 11))
print("arange step 2:", np.arange(2, 21, 2))
print("linspace:", np.linspace(0, 10, 5))
```

## 8. Random Numbers — Recommended Generator API

```python
rng = np.random.default_rng(42)

print("Random floats:", rng.random(5))
print("Random integers:", rng.integers(1, 100, size=10))
print("Normal distribution:", rng.normal(loc=0, scale=1, size=5))
print("Choice:", rng.choice(["AI", "ML", "DL"], size=5))
```

## 9. Indexing

```python
arr = np.array([10, 20, 30, 40, 50])

print("First:", arr[0])
print("Fourth:", arr[3])
print("Last:", arr[-1])

matrix = np.array([[10, 20, 30], [40, 50, 60]])
print("matrix[0, 1] =", matrix[0, 1])
print("matrix[1, 2] =", matrix[1, 2])
```

## 10. Slicing

```python
arr = np.arange(1, 11)
print(arr)
print("arr[2:6] =", arr[2:6])
lis = [1,2,3,4,5,6]
print(lis)
print("lis",lis[2:6])
print("arr[2:6] =", arr[2:6])
print("arr[:5] =", arr[:5])
print("arr[4:] =", arr[4:])
print("arr[::2] =", arr[::2])
print("reverse =", arr[::-1])

matrix = np.arange(1, 13).reshape(3, 4)
print("\nMatrix:\n", matrix)
print("\nFirst row:", matrix[0])
print("Second column:", matrix[:, 1])
print("Sub-matrix:\n", matrix[0:2, 1:3])
```

## 11. Reshape, Flatten, Ravel and Transpose

```python
arr = np.arange(1, 13)
matrix = arr.reshape(3, 4)

print(arr)
print("Reshaped:\n", matrix)
print("Flatten:", matrix.flatten())
print("Ravel:", matrix.ravel())
print("Transpose:\n", matrix.T)
```

## 12. reshape with -1

```python
arr = np.arange(1, 13)
print(arr)
print(arr.reshape(2, -1))
print(arr.reshape(-1, 3))
```

## 13. Element-wise Arithmetic

```python
a = np.array([1, 2, 3, 4])

print("a + 10:", a + 10)
print("a - 5:", a - 5)
print("a * 2:", a * 2)
print("a / 2:", a / 2)
print("a ** 2:", a ** 2)
```

## 14. Operations Between Arrays

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

print("Addition:", a + b)
print("Subtraction:", b - a)
print("Element-wise multiplication:", a * b)
print("Division:", b / a)
```

## 15. Universal Functions (ufuncs)

## 16. Aggregation Functions

```python
a = np.array([10, 20, 30, 40, 50])

print("sum:", np.sum(a))
print("mean:", np.mean(a))
print("median:", np.median(a))
print("min:", np.min(a))
print("max:", np.max(a))
print("std:", np.std(a))
print("variance:", np.var(a))
print("argmin:", np.argmin(a))
print("argmax:", np.argmax(a))
```

## 17. Aggregations by Axis

```python
m = np.array([[1, 2, 3], [4, 5, 6]])

print("Matrix:\n", m)
print("Column sums (axis=0):", m.sum(axis=0))
print("Row sums (axis=1):", m.sum(axis=1))
print("Column means:", m.mean(axis=0))
print("Row means:", m.mean(axis=1))
```

## 18. Broadcasting

```python
m = np.array([[1, 2, 3], [4, 5, 6]])
row = np.array([10, 20, 30])

print("Matrix + row vector:\n", m + row)

col = np.array([[100], [200]])
print("\nMatrix + column vector:\n", m + col)
```

## 19. Boolean Indexing

```python
a = np.array([10, 20, 30, 40, 50, 60])

mask = a > 30
print("Mask:", mask)
print("Values > 30:", a[mask])
print("Even and > 20:", a[(a % 2 == 0) & (a > 20)])
```

## 20. np.where and np.select

```python
scores = np.array([35, 55, 72, 88, 41])

result = np.where(scores >= 50, "Pass", "Fail")
print("where:", result)

conditions = [
    scores < 50,
    (scores >= 50) & (scores < 75),
    scores >= 75
]
choices = ["Fail", "Average", "Excellent"]

print("select:", np.select(conditions, choices, default="Unknown"))
```

## 21. Sorting and Searching

```python
a = np.array([50, 10, 40, 20, 30])

print("sort:", np.sort(a))
print("argsort:", np.argsort(a))
print("where a > 25:", np.where(a > 25)[0])
```

## 22. Unique Values

```python
a = np.array([1, 2, 2, 3, 3, 3, 4])

values, counts = np.unique(a, return_counts=True)
print("Unique:", values)
print("Counts:", counts)
```

## 23. Concatenate and Stack

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("concatenate:", np.concatenate((a, b)))
print("vstack:\n", np.vstack((a, b)))
print("hstack:", np.hstack((a, b)))
print("column_stack:\n", np.column_stack((a, b)))
```

## 24. Splitting Arrays

```python
a = np.arange(1, 13)

print("split into 3:", np.split(a, 3))
print("array_split into 5:", np.array_split(a, 5))
```

## 25. Copy vs View

```python
original = np.array([1, 2, 3, 4])
print(original)

copied = original.copy()  
viewed = original.view()

copied[0] = 100
viewed[1] = 200

print("Original:", original)
print("Copy:", copied)
print("View:", viewed)
```

## 26. Missing Values (NaN)

```python
a = np.array([10.0, np.nan, 30.0, np.nan, 50.0])

print(a)
print("isnan:", np.isnan(a))
print("Normal mean:", np.mean(a))
print("NaN-safe mean:", np.nanmean(a))
print("NaN-safe sum:", np.nansum(a))
print("Remove NaN:", a[~np.isnan(a)])
```

## 27. Matrix Multiplication

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Element-wise A * B:\n", A * B)
print("\nMatrix multiplication A @ B:\n", A @ B)
print("\nnp.matmul:\n", np.matmul(A, B))
```

## 28. Linear Algebra

```python
A = np.array([[4.0, 2.0], [1.0, 3.0]])
b = np.array([10.0, 8.0])

print("Determinant:", np.linalg.det(A))
print("Inverse:\n", np.linalg.inv(A))
print("Solve Ax=b:", np.linalg.solve(A, b))
print("Eigenvalues and eigenvectors:\n", np.linalg.eig(A))
print("Matrix norm:", np.linalg.norm(A))
```

## 29. Save and Load Arrays

```python
sample = np.array([[1, 2, 3], [4, 5, 6]])

np.save("numpy_sample.npy", sample)
loaded = np.load("numpy_sample.npy")

print(loaded)
```

## 30. Mini NumPy Exercise — Student Marks Analysis

```python
marks = np.array([
    [80, 75, 90],
    [65, 70, 72],
    [95, 88, 92],
    [50, 60, 55]
])

student_averages = marks.mean(axis=1)
subject_averages = marks.mean(axis=0)
top_student = np.argmax(student_averages)

print("Marks:\n", marks)
print("Student averages:", student_averages)
print("Subject averages:", subject_averages)
print("Top student index:", top_student)
print("Students with avg >= 75:", np.where(student_averages >= 75)[0])
```
