# Pandas Complete Implementation for AI/ML

Run the cells from top to bottom.

## 1. Import Pandas

```python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
```

## 2. Series

```python
marks = pd.Series([80, 90, 75, 88], name="Marks")
print(marks)

subject_marks = pd.Series(
    [80, 90, 75],
    index=["Maths", "Science", "English"],
    name="Marks"
)
print("\nCustom index:\n", subject_marks)
```

## 3. DataFrame

```python
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 106],
    "Name": [" Vineeth ", "Harshitha", "Priya", "Arjun", "Sneha", "Kiran", "Kiran"],
    "Department": ["AI", "HR", "AI", "Finance", "HR", "AI", "AI"],
    "Age": [25, 28, 24, 30, np.nan, 27, 27],
    "Salary": [70000, 50000, 65000, 60000, 52000, np.nan, np.nan],
    "Experience": [4, 5, 3, 6, 2, 4, 4],
    "City": ["Hyderabad", "Bangalore", "Hyderabad", "Chennai", "Bangalore", "Hyderabad", "Hyderabad"]
}

df = pd.DataFrame(data)
df
```

## 4. Save Sample Data and Read CSV

```python
df.to_csv("employees_sample.csv", index=False)
csv_df = pd.read_csv("employees_sample.csv")
csv_df.head()
```

## 5. Read / Write JSON

```python
df.to_json("employees_sample.json", orient="records", indent=2)
json_df = pd.read_json("employees_sample.json")
json_df.head()
```

## 6. Read / Write Excel

```python
try:
    df.to_excel("employees_sample.xlsx", index=False)
    excel_df = pd.read_excel("employees_sample.xlsx")
    print(excel_df.head())
except ImportError as e:
    print("Excel support requires openpyxl. Install it with: pip install openpyxl")
```

## 7. Inspect Data

```python
print("HEAD")
print(df.head(2))

print("\nTAIL")
print(df.tail(2))

print("\nSHAPE:", df.shape)
print("\nCOLUMNS:", df.columns.tolist())
print("\nDTYPES:\n", df.dtypes)

print("\nINFO")
df.info()

print("\nDESCRIBE")
print(df.describe())

print("\nDESCRIBE ALL")
print(df.describe(include="all"))
```

## 8. Selecting Columns

```python
print("Single column (Series):")
print(df["Name"])

print("\nMultiple columns (DataFrame):")
print(df[["Name", "Department", "Salary"]])
```

## 9. loc and iloc

```python
print("loc row 0:")
print(df.head())
print(df.loc[1])

print("\nloc rows 0:2, selected columns:")
print(df.loc[0:2, ["Name", "Salary"]])

print("\niloc first 3 rows, first 3 columns:")
print(df.iloc[0:3, 0:3])
```

## 10. Filtering

```python
print("Salary > 55000")
print(df[df["Salary"] > 55000])

print("\nAI employees")
print(df[df["Department"] == "AI"])

print("\nAI AND Experience >= 4")
print(df[(df["Department"] == "AI") & (df["Experience"] >= 4)])

print("\nHyderabad OR Chennai")
print(df[(df["City"] == "Hyderabad") | (df["City"] == "Chennai")])

print("\nisin")
print(df[df["Department"].isin(["AI", "Finance"])])

print("\nbetween")
print(df[df["Salary"].between(50000, 65000)])
```

## 11. Missing Values

```python
print("Missing matrix:\n", df.isnull())
print("\nMissing count:\n", df.isnull().sum())
print("\nNon-missing count:\n", df.notnull().sum())
```

## 12. Fill Missing Values

```python
df_clean = df.copy()

df_clean["Age"] = df_clean["Age"].fillna(df_clean["Age"].median())

dept_salary_mean = df_clean.groupby("Department")["Salary"].transform("mean")
df_clean["Salary"] = df_clean["Salary"].fillna(dept_salary_mean)
df_clean["Salary"] = df_clean["Salary"].fillna(df_clean["Salary"].median())

print(df_clean)
```

## 13. dropna, ffill and bfill

```python
print("dropna result:")
print(df.dropna())

print("\nForward fill:")
print(df.ffill())

print("\nBackward fill:")
print(df.bfill())
```

## 14. Duplicate Handling

```python
print("Duplicate mask:")
print(df.duplicated())

print("\nDuplicate count:", df.duplicated().sum())

df_no_dupes = df.drop_duplicates(subset=["Employee_ID"], keep="first")
print("\nAfter removing duplicate Employee_ID:")
print(df_no_dupes)
```

## 15. String Cleaning

```python
df_strings = df.copy()
df_strings["Name"] = df_strings["Name"].str.strip()

print("Uppercase names:")
print(df_strings["Name"].str.upper())

print("\nLowercase names:")
print(df_strings["Name"].str.lower())

print("\nName lengths:")
print(df_strings["Name"].str.len())

print("\nContains Hyderabad:")
print(df_strings[df_strings["City"].str.contains("hyderabad", case=False, na=False)])
```

## 16. Add New Columns

```python
work = df_clean.drop_duplicates(subset=["Employee_ID"]).copy()

work["Country"] = "India"
work["Annual_Salary"] = work["Salary"] * 12
work["Bonus"] = work["Salary"] * 0.10
work["Experience_Level"] = np.where(work["Experience"] >= 5, "Senior", "Junior")

work
```

## 17. Multiple Conditions with np.select

```python
conditions = [
    work["Experience"] < 3,
    work["Experience"].between(3, 5),
    work["Experience"] > 5
]
choices = ["Beginner", "Intermediate", "Senior"]

work["Level"] = np.select(conditions, choices, default="Unknown")
work[["Name", "Experience", "Level"]]
```

## 18. Update Values with loc

```python
updated = work.copy()
updated.loc[updated["Department"] == "AI", "Salary"] += 5000
updated[["Name", "Department", "Salary"]]
```

## 19. Rename and Drop

```python
renamed = work.rename(columns={"Employee_ID": "ID", "Department": "Dept"})
print(renamed.head())

dropped = work.drop(columns=["Country"])
print("\nAfter dropping Country:\n", dropped.head())
```

## 20. Sorting

```python
print(work.sort_values("Salary"))
print("\nDescending:\n", work.sort_values("Salary", ascending=False))
print("\nMultiple columns:\n", work.sort_values(["Department", "Salary"], ascending=[True, False]))
```

## 21. Aggregations

```python
salary = work["Salary"]

print("sum:", salary.sum())
print("mean:", salary.mean())
print("median:", salary.median())
print("min:", salary.min())
print("max:", salary.max())
print("std:", salary.std())
print("var:", salary.var())
print("count:", salary.count())

print("\nMultiple aggregations:")
print(salary.agg(["min", "max", "mean", "sum"]))
```

## 22. value_counts, unique and nunique

```python
print(work["Department"].value_counts())
print("\nPercentages:\n", work["Department"].value_counts(normalize=True) * 100)
print("\nUnique values:", work["Department"].unique())
print("Unique count:", work["Department"].nunique())
```

## 23. GroupBy

```python
print("Average salary by department:")
print(work.groupby("Department")["Salary"].mean())

print("\nMultiple aggregations:")
print(
    work.groupby("Department")["Salary"]
        .agg(["count", "mean", "min", "max"])
)

summary = (
    work.groupby("Department", as_index=False)
        .agg(
            Employee_Count=("Employee_ID", "count"),
            Average_Salary=("Salary", "mean"),
            Maximum_Salary=("Salary", "max"),
            Average_Experience=("Experience", "mean")
        )
)
summary
```

## 24. GroupBy transform

```python
tmp = work.copy()
tmp["Department_Avg_Salary"] = tmp.groupby("Department")["Salary"].transform("mean")
tmp[["Name", "Department", "Salary", "Department_Avg_Salary"]]
```

## 25. apply, lambda and map

```python
def salary_category(salary):
    if salary >= 65000:
        return "High"
    elif salary >= 55000:
        return "Medium"
    return "Low"

tmp = work.copy()
tmp["Salary_Category"] = tmp["Salary"].apply(salary_category)
tmp["Bonus_Vectorized"] = tmp["Salary"] * 0.10

department_codes = {"AI": 1, "HR": 2, "Finance": 3}
tmp["Department_Code"] = tmp["Department"].map(department_codes)

tmp[["Name", "Salary", "Salary_Category", "Department", "Department_Code"]]
```

## 26. Concatenation

```python
df1 = pd.DataFrame({"Name": ["A", "B"], "Salary": [50000, 60000]})
df2 = pd.DataFrame({"Name": ["C", "D"], "Salary": [55000, 65000]})
print(df1.head())
print(df2.head())
combined = pd.concat([df1, df2], ignore_index=True)
combined
```

## 27. Merge / SQL-style Joins

```python
employees = pd.DataFrame({
    "Employee_ID": [101, 102, 103],
    "Name": ["Vineeth", "Rahul", "Priya"]
})

departments = pd.DataFrame({
    "Employee_ID": [101, 102, 104],
    "Department": ["AI", "HR", "Finance"]
})

print("INNER")
print(pd.merge(employees, departments, on="Employee_ID", how="inner"))

print("\nLEFT")
print(pd.merge(employees, departments, on="Employee_ID", how="left"))

print("\nRIGHT")
print(pd.merge(employees, departments, on="Employee_ID", how="right"))

print("\nOUTER")
print(pd.merge(employees, departments, on="Employee_ID", how="outer"))
```

## 28. Date and Time Handling

```python
sales = pd.DataFrame({
    "Date": ["2026-08-01", "2026-08-02", "2026-08-03", "2026-09-01"],
    "Sales": [10000, 15000, 12000, 18000]
})

sales["Date"] = pd.to_datetime(sales["Date"])
sales["Year"] = sales["Date"].dt.year
sales["Month"] = sales["Date"].dt.month
sales["Day"] = sales["Date"].dt.day
sales["Day_Name"] = sales["Date"].dt.day_name()

sales
```

## 29. Pivot Table and Crosstab

```python

pivot = pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    columns="City",
    aggfunc="mean",
    fill_value=0
)
print("Pivot table:\n", pivot)

print("\nCrosstab:\n", pd.crosstab(work["Department"], work["City"]))
```

## 30. Set and Reset Index

```python
indexed = work.set_index("Employee_ID")
print(indexed.head())

reset = indexed.reset_index()
print("\nReset:\n", reset.head())
```

## 31. Correlation

```python
numeric_cols = work.select_dtypes(include="number")
print(numeric_cols.corr())
```

## 32. Rolling Window Example

```python
daily = pd.DataFrame({
    "Day": pd.date_range("2026-08-01", periods=7),
    "Sales": [100, 120, 90, 140, 160, 150, 180]
})

daily["3_Day_Moving_Average"] = daily["Sales"].rolling(3).mean()
daily
```

## 33. SQLite Integration

```python
import sqlite3

connection = sqlite3.connect(":memory:")
work.to_sql("employees", connection, index=False, if_exists="replace")

sql_result = pd.read_sql(
    "SELECT Department, AVG(Salary) AS AvgSalary FROM employees GROUP BY Department",
    connection
)

print(sql_result)
connection.close()
```
