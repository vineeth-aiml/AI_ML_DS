# Matplotlib Complete Implementation for AI/ML

* Run the cells from top to bottom.

## 1. Import

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Basic Line Plot

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 40]

plt.plot(x, y)
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Simple Line Chart")
plt.show()
```

## 3. Line Chart — Trend Over Time

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 130, 180, 220]

plt.plot(months, sales)
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")
plt.show()
```

## 4. Multiple Lines and Legend

```python
months = ["Jan", "Feb", "Mar", "Apr"]
product_a = [100, 120, 150, 170]
product_b = [80, 110, 140, 160]

plt.plot(months, product_a, label="Product A")
plt.plot(months, product_b, label="Product B")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Product Sales Comparison")
plt.legend()
plt.show()
```

## 5. Bar Chart

```python
departments = ["AI", "HR", "Finance", "Sales"]
employees = [20, 10, 8, 15]

plt.bar(departments, employees)
plt.xlabel("Department")
plt.ylabel("Employees")
plt.title("Employees by Department")
plt.show()
```

## 6. Horizontal Bar Chart

```python
plt.barh(departments, employees)
plt.xlabel("Employees")
plt.ylabel("Department")
plt.title("Employees by Department")
plt.show()
```

## 7. Scatter Plot

```python
experience = [1, 2, 3, 4, 5, 6]
salary = [30000, 35000, 42000, 50000, 60000, 70000]

plt.scatter(experience, salary)
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Experience vs Salary")
plt.show()
```

## 8. Histogram

```python
ages = [22, 23, 24, 25, 25, 26, 27, 27, 28, 30, 32, 35]

plt.hist(ages, bins=5)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.show()
```

## 9. Box Plot

```python
salary = [30000, 35000, 40000, 45000, 50000, 55000, 200000]

plt.boxplot(salary)
plt.ylabel("Salary")
plt.title("Salary Box Plot")
plt.show()
```

## 10. Pie Chart

```python
labels = ["AI", "HR", "Finance"]
values = [50, 30, 20]

plt.pie(values, labels=labels, autopct="%1.1f%%")
plt.title("Department Distribution")
plt.show()
```

## 11. Figure Size and Grid

```python
x = np.arange(1, 11)
y = x ** 2

plt.figure(figsize=(10, 5))
plt.plot(x, y)
plt.grid()
plt.xlabel("X")
plt.ylabel("X Squared")
plt.title("Figure Size and Grid")
plt.show()
```

## 12. Figure and Axes Object-Oriented API

```python
x = np.arange(1, 6)
y = x ** 2

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_xlabel("X")
ax.set_ylabel("X Squared")
ax.set_title("Object-Oriented Matplotlib")
plt.show()
```

## 13. Multiple Subplots

```python
x = np.arange(1, 6)

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].plot(x, x)
axes[0, 0].set_title("Linear")

axes[0, 1].plot(x, x ** 2)
axes[0, 1].set_title("Square")

axes[1, 0].bar(x, x * 2)
axes[1, 0].set_title("Bar")

axes[1, 1].scatter(x, x ** 3)
axes[1, 1].set_title("Scatter")

plt.tight_layout()
plt.show()
```

## 14. Axis Limits and Ticks

```python
x = np.arange(0, 11)
y = x ** 2

plt.plot(x, y)
plt.xlim(0, 10)
plt.ylim(0, 110)
plt.xticks(range(0, 11, 2))
plt.title("Axis Limits and Ticks")
plt.show()
```

## 15. Annotations

```python
x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 20, 18]

plt.plot(x, y)
plt.annotate(
    "Highest point",
    xy=(4, 20),
    xytext=(3, 22),
    arrowprops={"arrowstyle": "->"}
)
plt.title("Annotation Example")
plt.show()
```

## 16. Error Bars

```python
x = np.arange(1, 6)
means = np.array([10, 12, 15, 14, 18])
errors = np.array([1, 1.5, 0.8, 1.2, 1])

plt.errorbar(x, means, yerr=errors, capsize=4)
plt.xlabel("Experiment")
plt.ylabel("Mean Value")
plt.title("Error Bars")
plt.show()
```

## 17. Working Directly with Pandas

```python
df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Sales": [100, 140, 120, 180, 210]
})

plt.plot(df["Month"], df["Sales"])
plt.title("Pandas + Matplotlib")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

## 18. Save a Figure

```python
plt.plot(df["Month"], df["Sales"])
plt.title("Sales Chart")
plt.savefig("sales_chart.png", bbox_inches="tight")
plt.show()

print("Saved sales_chart.png")
```
